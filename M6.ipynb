```python
usia = int(input("Masukkan usia: "))
departemen = input("Masukkan nama departemen: ")

syarat_usia = usia >= 19
syarat_departemen = departemen.find("Statistika")

hasil = syarat_usia and syarat_departemen
print(f"Formulir terpenuhi: {hasil}")
```
